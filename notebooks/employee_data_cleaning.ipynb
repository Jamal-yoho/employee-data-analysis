```python
df=df.drop_duplicates()
print("Duplicate Rows After Cleaning:",df.duplicated().sum())
```

```python
df.dtypes
```

```python
df.info()
```

```python
text_columns = df.select_dtypes(include="str").columns

for column in text_columns:
    df[column] = df[column].str.strip()

print("Text standardisation completed.")
```

```python
for column in text_columns:
    print(column,df[column].str.contains(r"^\s|\s$", regex=True).sum())
```

```python
numeric_columns = df.select_dtypes(include = "number").columns
Q1 = df[numeric_columns].quantile(0.25)
Q3 = df[numeric_columns].quantile(0.75)

IQR = Q3 -Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = ((df[numeric_columns]<lower_bound) | (df[numeric_columns]> upper_bound))

print("Outliers in Each Column:")
print(outliers.sum())

print("\nTotal Outliers:", outliers.sum().sum())
```

## Outlier Analysis

Outliers were detected using the Interquartile Range (IQR)method.
The IQR method was used to identify values below the lower bound and above the upper bound.
A total of 1024 potential outlier values were identified across the numerical columns.
These values were retained because they may represent valid employee information. No outliers were removed. 


```python
before_rows = 1470
before_missing = 0 
before_duplicates = 0

print("Before Cleaning Summary")
print("Rows: ", before_rows)
print("Missing Values: ", before_missing)
print("Duplicates: ", before_duplicates)
```

```python
after_rows = df.shape[0]
after_missing = df.isna().sum().sum()
after_duplicates = df.duplicated().sum()

print("After Cleaning Summary")
print("Rows: ",after_rows)
print("Missing Values: ", after_missing)
print("Duplicates: ", after_duplicates)
```

```python
print("Missing Values: ", df.isna().sum().sum())
```

```python
summary = pd.DataFrame({
    "Metric": ["Rows", "Missing Values", "Duplicates"],
    "Before Cleaning": [before_rows,before_missing,before_duplicates],
    "Ater Cleaning": [after_rows,after_missing,after_duplicates]
})

summary
```
