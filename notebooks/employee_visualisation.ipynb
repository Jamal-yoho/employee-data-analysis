# Employee Data Visualisation

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load Cleaned Dataset

```python
df= pd.read_csv("../data/clean_data.csv")
df.head()
```

## 3. Understanding the Dataset

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

## 4. Bar Chart - Employee Count by Department

```python
department_count = df["Department"].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(department_count.index, department_count.values,color="skyblue",label="Employees")
plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.legend()

plt.tight_layout()
plt.savefig("../reports/figures/bar_chart.png")
plt.show()
```

**Insight:** THe chart shows the distribution of employees across different departments.

## 5. Histogram - Employee Age Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"],bins=10, color="skyblue",edgecolor="black",label="Employees")
plt.title("Employee Age Distribution")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Employees")

plt.legend()
plt.tight_layout()
plt.savefig("../reports/figures/histogram.png")
plt.show()
```

**Insights:** The histogram shows how employees are distributed across different age groups.

## 6. Scatter Plot - Age vs Monthly Income

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Age"],df["MonthlyIncome"], color="orange", label="Employees")

plt.title("Employee Age vs MonthlyIncomeary")
plt.xlabel("Age(Years)")
plt.ylabel("MonthlyIncome (Currency Units)")
plt.legend()
plt.tight_layout()
plt.savefig("../reports/figures/scatter_plot.png")
plt.show()
```

**Insight:** The scatter plot shows the relationship between employee age and monthly income

## 7. Line Chart - Average Monthly Income by Age 

```python
age_income = df.groupby("Age")["MonthlyIncome"].mean()

plt.figure(figsize=(8,5))
plt.plot(age_income.index,age_income.values,color="green", marker="o", label="Average Income")
plt.title("Average Monthly Income by Age")
plt.xlabel("Age(Years)")
plt.ylabel("Average Monthly Income (Currency Units)")
plt.legend()
plt.tight_layout()
plt.savefig("../reports/figures/line_chart.png")
plt.show()
```

**Insight:** The line chart shows how average monthly income changes across different employee ages.

## 8. Box Plot - Monthly Income Distribution

```python
plt.figure(figsize=(8,5))
plt.boxplot(df["MonthlyIncome"],patch_artist=True, boxprops=dict(facecolor="skyblue"),label=["Monthlt Income"])
plt.title("Monthly Income Distribution")
plt.xlabel("Employee Income")
plt.ylabel("Monthly Income (Currency Units)")
plt.tight_layout()
plt.savefig("../reports/figures/box_plot.png")
plt.show()
```

**Insight:** The box plot shows the spread of employee monthly income and highlights potential outliers.
